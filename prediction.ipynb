```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
```

```python
model = load_model('model.h5')
```

```python
with open('geography_ohe.pkl', 'rb') as f:
    geography_ohe = pickle.load(f)
```

```python
with open('gender_le.pkl', 'rb') as f:
    gender_le = pickle.load(f)
```

```python
with open('ss.pkl', 'rb') as f:
    ss = pickle.load(f)
```

```python
input_data = {
    'CreditScore':600,
    'Geography':'France',
    'Gender':'Male',
    'Age':40,
    'Tenure':3,
    'Balance':60000,
    'NumOfProducts':2,
    'HasCrCard':1,
    'IsActiveMember':1,
    'EstimatedSalary':50000
}
```

```python
geography = geography_ohe.transform([[input_data['Geography']]]).toarray()
geo_col = geography_ohe.get_feature_names_out(['Geography'])
df_geography = pd.DataFrame(geography, columns=geo_col)
```

```python
df_geography
```

```python
input_data_df = pd.DataFrame([input_data])
```

```python
input_data_df
```

```python
input_data_df.drop(['Geography'], axis=1, inplace=True)
```

```python
input_data_df
```

```python
input_data_df = pd.concat([input_data_df,df_geography],axis=1)
```

```python
input_data_df
```

```python
gen = gender_le.transform(input_data_df['Gender'])
gen
```

```python
input_data_df['Gender'] = gen
input_data_df
```

```python
input_scaled = ss.transform(input_data_df)
input_scaled
```

```python
prediction = model.predict(input_scaled)
prediction
```

```python
prediction_probablity = prediction[0][0]
```

```python
if prediction_probablity>0.5:
    print("Customer is likely to churn.")
else:
    print("Customer is not likely to churn.")
```
